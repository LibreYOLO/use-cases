## Detect objects

The weights download from Hugging Face the first time you use them.

```python
from IPython.display import Image, display
from libreyolo import LibreYOLO

model = LibreYOLO("LibreYOLO9t.pt")
image = "https://raw.githubusercontent.com/LibreYOLO/libreyolo/release/libreyolo/assets/guggenheim-bilbao.jpg"
result = model(image, save=True)
display(Image(filename=result.saved_path))
```

```python
for box, cls, conf in zip(result.boxes.xyxy, result.boxes.cls, result.boxes.conf):
    print(result.names[int(cls)], f"{float(conf):.2f}", [round(float(v)) for v in box])
```

To try your own picture, set `image` to a URL or to a file you uploaded in the Files panel, and run the cell again.

Other models load the same way: `LibreYOLO9s.pt` or `LibreYOLO9m.pt` for more accuracy, and `LibreRFDETRn.pt` for RF-DETR (install `libreyolo[rfdetr]` first). All of them are listed in the [model docs](https://www.libreyolo.com/docs/models).

## Export to ONNX

The exported file loads with the same call and runs with ONNX Runtime. The ONNX simplifier prints a few `WARNING: failed to run ... op` lines while exporting; they are harmless.

```python
onnx_path = model.export(format="onnx")
onnx_model = LibreYOLO(str(onnx_path))
display(Image(filename=onnx_model(image, save=True).saved_path))
```
